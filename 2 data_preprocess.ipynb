```python
import csv

def csv_to_dict(file_path):
    # Initialize an empty dictionary
    result_dict = {}

    # Open the CSV file
    with open(file_path, mode='r', newline='', encoding='utf-8') as file:
        # Create a CSV reader
        csv_reader = csv.reader(file)

        # Skip the header if present
        next(csv_reader, None)  # Uncomment this line if your CSV has a header row

        # Iterate through the rows and add entries to the dictionary
        for row in csv_reader:
            key = f'{row[0]}_task4'  # Column A (index 0)
            value = row[1:]# Column B (index 1)
            result_dict[key] = value

    return result_dict

# Example usage
file_path = 'task4_preprocessed/MAT_task4_preprocessed/Statistics/Dimensions_task4_preprocessed.csv'
data = csv_to_dict(file_path)
print(data)
```

```python
for speech in data:
    data[speech].append("task4")
    if "China" in speech:
        data[speech].append("China")
    elif "United States" in speech:
        data[speech].append("UnitedStates")
    elif "no_nationality" in speech:
        data[speech].append("NoNationality")
    else:
        data[speech].append("NA")

print(data)
```

```python
def save_dict_to_csv(data, file_path):
    # Open the CSV file for writing
    with open(file_path, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)

        # Write the header row
        writer.writerow(['file', 'd1', 'd2', 'd3', 'd4', 'd5', 'd6', 'type', 'task', 'nationality'])

        # Write the data rows
        for key, values in data.items():
            row = [key] + values  # Combine the key with the values list
            writer.writerow(row)

# Example usage
file_path = 'statistics_4.csv'
save_dict_to_csv(data, file_path)
print(f"Data has been saved to {file_path}")
```
