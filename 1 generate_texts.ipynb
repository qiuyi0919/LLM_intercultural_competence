```python
from openai import OpenAI
```

```python
# Load OpenAI API client

API_KEY = "insert api key"
BASE_URL = "https://api.deepbricks.ai/v1/"
client = OpenAI(api_key=API_KEY, base_url=BASE_URL)
```

```python
nationalities = ['the United States', 'China']
```

```python
def task_1_1 (nationality):
    # Define the prompt
    prompt = (
            f"You are an employee working at a global company. A coworker on your team, who is from {nationality}, has repeatedly missed deadlines, delaying the project. Write a 300-word email to the coworker to address this issue. Write the email in a way that aligns with the coworker’s cultural values."
                )
    response = client.chat.completions.create(
            model="gpt-4o",
            messages=[
                        {
                            "role": "user",
                            "content": prompt
                        },
                    ],
                    temperature = 1,
                )

    answer = response.choices[0].message.content
    return answer               
```
